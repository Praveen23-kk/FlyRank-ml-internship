## 1. My lane (or freestyle) and why

**Lane: Lead Quality / High-Value Ranking.**

**Why:** FlyRank's current system treats every lead with equal urgency, leading to sales fatigue and missed opportunities on high-intent signups. By choosing this lane, I aim to create a score that differentiates between a 'curiosity' signup and a 'high-intent' enterprise signup, ensuring human effort is applied where the expected value is highest.

```python
import pandas as pd

# Load the data if not already present
url = 'https://raw.githubusercontent.com/applied-ml-internship/flyrank-datasets/main/flyrank_starter.csv'
try:
    df = pd.read_csv(url)
except:
    # Fallback to local if repo is unreachable during execution
    pass

# Basic stats to justify the lane
if 'df' in locals():
    total_leads = len(df)
    conversions = df['converted'].sum()
    print(f"Total leads in starter set: {total_leads}")
    print(f"Baseline conversions: {conversions}")
    print(f"Conversion Rate: {(conversions/total_leads)*100:.2f}%")
    print("\nThis low baseline (approx 12.5%) justifies a ranking model to filter the 87.5% of noise.")
```

## 2. The question: decision, action, cost of a wrong call

**Decision:** Should a sales representative call this lead immediately, or move it to a lower-priority automated nurturing sequence?

**Action:** The CRM will sort the daily lead list by 'Predicted Conversion Score' instead of 'Signup Date'.

**Cost of a wrong call:** A 'False Positive' (high score, no conversion) wastes expensive human sales hours. A 'False Negative' (low score, would have converted) results in a lost customer because a competitor reached them first while they sat in our low-priority queue.

```python
# Quantifying the 'Cost of a wrong call'
# If we have 4000 leads and only ~500 convert, we are looking for a 'needle in a haystack'.
# A simple check on the volume of non-converters we are currently calling:
if 'df' in locals():
    non_converters = len(df[df['converted'] == 0])
    print(f"Current 'Noise' (Leads that won't convert): {non_converters}")
    print(f"Potential time saved if we deprioritize even 20% of noise: {non_converters * 0.2:.0f} leads not called manually.")
```

## 4. Careful words: what I can and can't claim

**What I can claim:** This model provides a **directional decision-support** tool. We can measure if leads ranked in the top 10% convert at a statistically higher rate than the bottom 10% in historical data (**observed**).

**What I can't claim:** I cannot claim **causal proof** (e.g., that the ranking *caused* the conversion). I also cannot 'predict Google' or external market shifts that might change lead behavior overnight.

```python
# Verifying the available features for directional claims
if 'df' in locals():
    print("Available features for directional scoring:")
    print(df.columns.tolist())
    print("\nBy observing correlations between these features and 'converted', we can provide support,")
    print("but without a randomized controlled trial, we cannot claim causation.")
```
